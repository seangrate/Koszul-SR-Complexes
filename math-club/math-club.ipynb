# Simplicial complexes forcing failure of the weak Lefschetz property

**Definition**: A *simplicial complex* $\Delta$ on the *vertex set* $[n] = \{ 1,\dots,n \}$ is a collection of subsets called *faces* or *simplices*, closed under taking subsets.

```python
import sys
sys.path.append('../')

from simplicial import SimplicialComplex


# Make a simplicial complex
delta = SimplicialComplex([{1,2,3,4},{1,5}])
print(f'Faces: {delta.maximal_faces}')
delta.draw(shade=True, with_labels=True)

# Some statistics of the simplicial complex
print(f'Dimension: {delta.dim}\n'
      f'is pure: {delta.is_pure}\n'
      f'f-vector: {delta.f_vector}\n'
      f'h-vector: {delta.h_vector}\n'
      f'Betti numbers: {delta.betti_numbers}')
```

```python
# Make the hollow octahedron
delta = SimplicialComplex([{1,2,5},{2,3,5},{3,4,5},{1,4,5},
                           {1,2,6},{2,3,6},{3,4,6},{1,4,6}])
delta.draw(shade=True, with_labels=True)

# Some statistics of the simplicial complex
print(f'Dimension: {delta.dim}\n'
      f'is pure: {delta.is_pure}\n'
      f'f-vector: {delta.f_vector}\n'
      f'h-vector: {delta.h_vector}\n'
      f'Betti numbers: {delta.betti_numbers}')
```

# Key Ideas
- To record the data of a simplicial complex, it suffices to know the maximal faces (equivalently, the minimal non-faces).
- We can encode this data by thinking of the vertices as labeled by variables, and then equating non-faces to be polynomials which are set to zero.

**Definition**:
Let $\Delta$ be a simplicial complex on $[n]$. The **Stanley-Reisner ideal* of $\Delta$ is the ideal $I_\Delta \subseteq R = \mathbb{K}[x_1,\dots,x_n]$ given by
$$I_\Delta = \left( \prod_{i \in \sigma} x_i \, | \, \sigma \not\in \Delta \right).$$
The *Stanley-Reisner ring* (or *face ring*) of $\Delta$ is the quotient ring $R / I_\Delta$.

# Some algebra
**Definition**: Let $R = \mathbb{K}[x_1,\dots,x_n]$ and $I \subseteq R$ be an ideal. The quotient ring $A = R/I$ is called *Artinian* if it is a finite-dimensional vector space over $\mathbb{K}$. 

In other words, $A$ is Artinian if $A_d = 0$ for $d \gg 0$.

<br>
<br>
<br>

**Example**: $A = \mathbb{K}[x,y,z] / \langle x^2,y^2,z^3 \rangle$

| $d$ || $0$ | $1$ | $2$ | $3$ | $4$ | $5$ |
| :---: | :---: | :---: | :---: | :---: | :---: | :---: | :---: |
| $\text{gens}(A_d)$ || $1$ | $x$<br>$y$<br>$z$ | $xy$<br>$xz$<br>$yz$<br>$z^2$ | $xyz$<br>$xz^2$<br>$yz^2$ | $xyz^2$ | $\emptyset$ |
| $\text{dim}(A_d)$ || $1$ | $3$ | $4$ | $3$ | $1$ | $0$ |

<br>

**Example**: If $I_\Delta \subseteq \mathbb{K}[x_1,\dots,x_n]$ is a Stanley-Reisner ideal, then $I_\Delta + \langle x_i^2 \, | \, 1 \leq i \leq n \rangle$ is Artinian.

## Weak Lefschetz property (WLP)

For all $f \in A$, $f$ is a zero-divisor.
The best we can hope for is $\exists \, f \in A$ such that $\cdot f$ is injective or surjective in all degrees.

**Definition**: $A = R/I$ has the *weak Lefschetz property* (*WLP*) if $\times \ell \colon A_i \to A_{i+1}$ is full rank (i.e., either injective or surjective) for all $i \geq 0$.

## WLP is hard

<br>

**Theorem** [Brenner, Kaid; 2010]

Let $\text{char}(\mathbb{K}) = 2$. Then $A = \mathbb{K}[x,y,z] / \langle x^d, y^d, z^d \rangle$ has the WLP if and only if $d = \left\lfloor \frac{2^k + 1}{3} \right\rfloor$ for some positive integer $k$.

<br>

**Theorem** [Harbourne, Schenck, Seceleanu; 2011]

Let
$$ I = \langle L_1^t, \dots, L_n^t \rangle \subset \mathbb{K}[x_1,x_2,x_3,x_4] $$
with $L_i \in R_1$ generic. If $n \in \{5,6,7,8\}$, then the WLP fails, respectively, for $t \geq \{3,27,140,704\}$.

## Betti tables

Given an ideal $I \subset \mathbb{K}[x_1,\dots,x_n]$, a (minimal) free resolution "approximates" $R/I$. The betti table records the ranks of the summands appearing in the free resolution.

**Example**: $R = \mathbb{K}[x,y,z]$ and $I = \langle x^2, y^2, z^3 \rangle$
$$0 \leftarrow R/I \leftarrow R \leftarrow \substack{R(-2)^2 \\ \oplus \\ R(-3)} \leftarrow \substack{R(-4) \\ \oplus \\ R(-5)^2} \leftarrow R(-7) \leftarrow 0$$

# Strategy

**Definition**: A Betti table $B$ has an *$(n,d)$-Koszul tail* if it has an upper-left principal block of the form
$$
\begin{array}{c|ccccccccc}
     & 0 & 1 & 2 & 3 & \dots & n-2 & n-1 & n & n+1 \\
    \hline
    0 & 1 & . & . & . & \dots & . & . & . & * \\
    1 & . & . & . & . & \dots & . & . & . & * \\
    \vdots & \vdots & \vdots & \vdots & \vdots & \ddots & \vdots & \vdots & \vdots & * \\
    d-1 & . & . & . & . & \dots & . & . & . & * \\
    d & . & n & \binom{n}{2} & \binom{n}{3} & \dots & \binom{n}{n-2} & n & 1 & * \\
    d+1 & * & * & * & * & * & * & * & * & *
\end{array}.
$$
If $B$ is has an $(n,d)$-Koszul tail and is the Betti table for an Artinian ring $\mathbb{K}[x_1,\dots,x_n]/I$, then we say $B$ has a *maximal*  $(n,d)$-Koszul tail.

<br>

**Theorem** [Grate, Schenck; 2024]: If an Artinian algebra $A$ has a maximal Koszul tail, then $A$ does not have the WLP.
